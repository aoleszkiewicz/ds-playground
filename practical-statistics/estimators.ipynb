```python
import numpy as np
```

```python
data = np.linspace(1, 100, 30)
```

### Location estimators

```python
def mean(data):
    return np.sum(data) / len(data)
```

```python
print(mean(data), np.mean(data))
```

```python
def median(data):
    n = len(data)
    if n % 2 != 0:
        index = int(n / 2)
        return data[index]
    else:
        index_1 = int((n / 2) - 1)
        index_2 = int(n / 2)
        return (data[index_1] + data[index_2]) / 2
```

```python
print(median(data), np.median(data))
```

```python
data_with_outliers = data.copy()
data_with_outliers[0] = -1000
data_with_outliers[-1] = 1000
data_with_outliers
```

```python
print(f"mean: {mean(data_with_outliers)} \n median: {median(data_with_outliers)}")
```

```python
def weighted_mean(data, weights):
    if len(weights) != len(data):
        raise ValueError("weights lenth must match data's lenth")
    
    x_w = sum(weights * data) / sum(weights)
    return x_w
```

```python
weights = np.random.randn(*data.shape)

print(weighted_mean(data, weights))
```

```python
def trimmed_average(p, data):
    n = len(data)
    acc = 0
    for i in range(p, n - p):
        acc += data[i]

    return acc / (n - 2*p)
```

```python
print(trimmed_average(2, data))
```

### Measures of variation

$\sum_(x_i - \hat{x})^2 \over{n-1}$

```python
def variation(data):
    n = len(data)
    return (sum(data - mean(data)) ** 2) / (n - 1)
```

```python
print(variation(data))
```
