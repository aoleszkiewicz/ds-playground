```python
import numpy as np


def train_test_split_custom(X, y, test_size=0.2, random_state: int | None = None):
    # 1. Walidacja
    if X.shape[0] != y.shape[0]:
        raise ValueError("Rows does not match")
    if test_size < 0.0 or test_size > 1.0:
        raise ValueError(
            "test_size cannot be greater than 1.0 or less than 0.0 (negative)"
        )
    if random_state is not None and not isinstance(random_state, int):
        raise ValueError("random_state must be an integer")

    # 2. Setup generatora losowego (deterministyczny jeśli random_state podany)
    rng = np.random.default_rng(random_state)

    # 3. Permutacja indeksów [0, 1, ..., n-1]
    n_samples = X.shape[0]
    indices = rng.permutation(n_samples)

    # 4. Obliczenie punktu cięcia
    n_test = int(n_samples * test_size)
    n_train = n_samples - n_test

    # 5. Podział indeksów na train/test
    train_indices = indices[:n_train]
    test_indices = indices[n_train:]

    # 6. Indeksowanie X i y
    X_train = X[train_indices]
    y_train = y[train_indices]
    X_test = X[test_indices]
    y_test = y[test_indices]

    # 7. Return
    return X_train, X_test, y_train, y_test
```

```python
X = np.arange(20).reshape(10, 2)
y = np.arange(10)

X_train, X_test, y_train, y_test = train_test_split_custom(
    X, y, test_size=0.2, random_state=42
)

assert X_train.shape == (8, 2)
assert X_test.shape == (2, 2)

# Determinism check
X_train2, _, _, _ = train_test_split_custom(X, y, test_size=0.2, random_state=42)
assert np.array_equal(X_train, X_train2), "Not deterministic!"

# X-y alignment check
for x_row, y_val in zip(X_train, y_train):
    assert x_row[0] == 2 * y_val, "X and y misaligned!"

print("All sanity checks passed ✅")
```
